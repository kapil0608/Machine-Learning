```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df=pd.read_csv("spam.csv")
df
```

```python
df.info()
```

```python
x = df[['v2']]
y = df['v1']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
xtrain
```

```python
vectorization = CountVectorizer(stop_words='english',binary=True,min_df=1,max_df=1.0,max_features=100,ngram_range=(2,2))
vectorization
```

```python
vectors =vectorization.fit_transform(xtrain['v2'])         #[['v2']] => 'v2' => 1
vectors
```

```python
vectors.toarray()
```
