```python
import pandas as pd
from sklearn.impute import SimpleImputer
import numpy as np
```

```python
df=pd.DataFrame(
    data = {
    "Age": [25, 30, 22, np.nan, 28],
    "Number": [4564, 1566, np.nan, 4856, np.nan],
    "City": ["Delhi", "Mumbai", "Delhi", "Bengaluru", np.nan]
    }
)
df
```

```python
import seaborn as sns
sns.boxplot(df)          # 0 Outliers
```

```python
imputer=SimpleImputer(strategy='constant',fill_value=0)
```

```python
imputer.fit_transform(df)
```

```python
imputer.fit_transform(df[['Age','Number']])
```

```python
imputer=SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Age','Number']])

# imputer=SimpleImputer(strategy='mean')
# imputer.fit_transform(df[['Age','Number','City']])      # It will give ValueError ,because we can not use categorical column
```

```python
imputer=SimpleImputer(strategy='median')
imputer.fit_transform(df[['Age','Number']])
```

```python
imputer=SimpleImputer(strategy='most_frequent')
imputer.fit_transform(df[['Age','Number','City']])      # When multiple values have the same frequency, assign the most frequent value; otherwise, assign the first value in ascending order.
```
