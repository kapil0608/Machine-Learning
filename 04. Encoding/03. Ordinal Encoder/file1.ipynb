```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
    "Category" : ["Electronics","Clothing","Grocery","Electronics","Clothing","Furniture","Grocery","Electronics","Furniture","Clothing"],
    "Brand" : ["Samsung","Nike","Nestle","Apple","Adidas","IKEA","Amul","Sony","Godrej","Puma"],
    "target": ["class0","class1","class2","class3","class2","class1","class2","class1","class2","class1"]
})
df
```

```python
x=df.drop(columns='target')
y=df['target']
```

```python
xtrain,xtest,ytain,ytest= train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
xtrain
```

```python
xtrain['Category'].unique()
```

```python
xtrain['Brand'].unique()
```

```python
encoder = OrdinalEncoder(
    categories=[['Furniture','Electronics','Grocery','Clothing'],["IKEA","Samsung","Sony","Nestle","Puma","Adidas","Apple","Amul"]],
    handle_unknown='use_encoded_value',
    unknown_value=-1
)

encoder.fit_transform(xtrain)
```

```python
encoder.transform(xtest)
```

```python
new_input=[["Salt",'TATA']]
```

```python
encoder.transform(new_input)
```
