```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame(
    data={
        "size":["small","small","medium","large","large"]
    }
)
df
```

```python
x=df['size']
y=df['size']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
xtrain
```

```python
df['size'].unique()
```

```python
encoder= OrdinalEncoder(
    # size=['small', 'medium', 'large']            # TypeError: OrdinalEncoder.__init__() got an unexpected keyword argument 'size'
    
    # categories= ['small', 'medium', 'large'],      # ValueError: Shape mismatch: if categories is an array, it has to be of shape (n_features,).

    categories= [['small', 'medium', 'large']],
    handle_unknown="use_encoded_value" ,
    unknown_value=-1
)
```

```python
# encoder.fit_transform(xtrain)
'''
ValueError: Expected a 2-dimensional container but got <class 'pandas.Series'> instead. Pass a DataFrame containing a single row (i.e. single sample) or a single column (i.e. single feature) instead.
'''

encoder.fit_transform(xtrain.to_frame())
```
