```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame(
    {
    "color": ["red","blue","green","blue","red","green","green","green","red","blue"],
    "size": ["S","M","L","XL","S","L","S","M","XL","M"],
    "shape": ["circle","square","triangle","circle","square","triangle","triangle","circle","square","circle"],
    "area" : [25, 36, 38, 25.8, 85, 45, 65, 74, 20, 41],
    "target": ["class1","class2","class1","class2","class1","class2","class1","class2","class1","class2"]
    }
)
df
```

```python
x=df.drop(columns='target')
y=df['target']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
encoder= OrdinalEncoder(
    categories=[['red', 'blue', 'green'],['S', 'M', 'L', 'XL'],['circle', 'square', 'triangle']],
    handle_unknown='use_encoded_value',
    unknown_value=-1,
)

encoder.fit_transform(xtrain[['color','size','shape']])         # accept 2D input
```

```python
encoder.fit_transform(xtest[['color','size','shape']]) 
```

```python
new_input= [['green','XS','circle']]
print(encoder.fit_transform(new_input))
encoder.transform(new_input)
```

```python
new_input=[['Yellow','XXL','circle']]
print(encoder.fit_transform(new_input))
print(encoder.transform(new_input))
```
