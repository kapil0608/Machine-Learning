```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
df=pd.DataFrame(
    {
    "color":["Blue","Red","Red","Blue"],
    "product_type": ["wood","plastic","wood","wood"],
    "label": ["class0","calss1","class2","class0"]
}
)
df
```

***Assigning x,y***

```python
x=df.drop(columns='label')
y= df['label']
```

***Apply Train_test_split***

```python
xtrain,xtest,ytrain,ytest= train_test_split(x,y,train_size=0.8, random_state=42)
```

```python
xtrain
```

***Model creation***

```python
encoder=OneHotEncoder(handle_unknown='ignore', sparse_output=False)
```

```python
xtrain_encoder= encoder.fit_transform(xtrain[['color','product_type']])
xtest_encoder= encoder.transform(xtest[['color','product_type']])
```

***Scenario-1***

```python
new_input=[['Blue','wood']]
encoder.transform(new_input)
```

***Scenario-2***

```python
new_input=[['Blue',"Plastic"]]
encoder.transform(new_input)
```
