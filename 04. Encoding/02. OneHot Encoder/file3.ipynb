```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame(data={
    "color": ["red","blue","green","blue","red","green","green","green","red","blue"],
    "size": ["S","M","L","XL","S","L","S","M","XL","M"],
    "shape": ["circle","square","triangle","circle","square","triangle","triangle","circle","square","circle"],
    "target": ["class1","class2","class1","class2","class1","class2","class1","class2","class1","class2"]
})
df
```

```python
x=df.drop(columns='target')
y=df['target']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8, random_state=42)
```

```python
xtrain
```

```python
encoder= OneHotEncoder()
encoder.fit_transform(xtrain[['color','size','shape']])
```

```python
print(encoder.get_feature_names_out())
print(type(encoder.fit_transform(xtrain[['color','size','shape']])))
```
