***We use only categorical data***

```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder
```

```python
data={
    "Brand" : ["Samsung","Apple","Nike","Sony","Philips","Adidas","Prestige","Dell","Samsung","Apple"]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
encoder = LabelEncoder()
```

```python
df=encoder.fit_transform(df['Brand'])
```
